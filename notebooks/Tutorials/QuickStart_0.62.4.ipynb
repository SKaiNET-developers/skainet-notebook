# SKaiNET Kotlin Notebook — Quick Start (0.62.4)

This notebook demonstrates the `sk.ainet.app:kotlin-notebook:0.62.4` artifact published to your local Maven repository. Run the cells top-to-bottom.

```kotlin
USE {
    repositories {
        mavenLocal()
    }
    dependencies {
        implementation("sk.ainet.app:kotlin-notebook:0.62.4")
    }
}
```

```kotlin
import sk.ainet.app.notebook.*
import sk.ainet.app.notebook.display.*
import sk.ainet.app.notebook.tools.*
import java.awt.Color
import java.awt.image.BufferedImage

NotebookInfo.info()
```

```kotlin
// Display a single image with options
val img = BufferedImage(128, 128, BufferedImage.TYPE_INT_RGB).apply {
    createGraphics().run {
        color = Color(0x00, 0xFF, 0x00)
        fillRect(0, 0, 128, 128)
        dispose()
    }
}
display(img) {
    width = 196
    height = 196
    alt = "green square"
    border = true
}
```

```kotlin
// Multi-image grid
val imgs = List(6) { idx ->
    BufferedImage(80, 80, BufferedImage.TYPE_INT_RGB).apply {
        createGraphics().run {
            color = Color((40 + 30 * idx) % 255, 150, 180)
            fillRect(0, 0, 80, 80)
            dispose()
        }
    }
}
display(imgs) { border = true }
```

```kotlin
// display(...) accepts URL strings, java.nio.file.Path, java.net.URL, ByteArray, or BufferedImage
display("https://picsum.photos/256")
```
